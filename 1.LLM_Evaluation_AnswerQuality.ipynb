# Checking the quality of LLM Answers

```python
import google.generativeai as genai
```

```python
genai.configure(api_key = "YOUR_API_KEY_HERE")
aimodel   = "gemini-3.5-flash-lite"
```

```python
# AI TEACHER INSTRUCTIONS

TEACHER_INSTRUCTIONS = """

You are an expert AI Teacher. Your job is to teach students by explaining concepts rather than just providing direct answers. 

Follow these strict rules:
1. Explain the concept in simple, layman's terms.
2. Limit your entire response to exactly 3 lines of text.
3. Formatting: Provide only the simple explanation. Do not include introductory text, conversational filler, or headers.
"""
```

```python
# CONVERSATION MEMORY
conversation = [
    {
    "role"    : "user",
    "parts" : TEACHER_INSTRUCTIONS
    }
]
```

```python
# AI TEACHER FUNCTION

def ask_teacher(question, subject = "AI & Data Science", student_level = "Beginner", language = "English"):
    
    user_message = f""" 
        Subject            : {subject}
        Student Level      : {student_level}
        Preferred Language : {language}
        Student Question   : {question} """

    #print(conversation)        
    conversation.append({"role": "user", "parts" : user_message})
    #print(conversation)

    model    = genai.GenerativeModel(aimodel)
    response = model.generate_content(conversation, generation_config = {"temperature" : 0})
    
    answer = response.text
    print(answer)

    conversation.append({"role" : "assistant", "parts" : answer })
    
```

```python
# CHAT APPLICATION

def start_teacher():

    print("=" * 60)
    print("              AI TEACHER")
    print("=" * 60)

    print("Type 'exit' to stop.\n")

    while True:
        question = input("Student: ")

        if question.lower() == "exit":
            print("\nTeacher: Keep learning!")
            break

        answer = ask_teacher(question = question, subject = "AI & Data Science", student_level = "Beginner", language = "English")

        print("\nTeacher:")
        print(answer)

        print("\n" + "-" * 60)
            
```

```python
# RUN

if __name__ == "__main__":
    start_teacher()
```
